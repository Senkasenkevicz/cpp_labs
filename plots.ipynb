```python
import matplotlib.pyplot as plt
import glob
import numpy as np

```

```python
file_pattern = '0tr*.txt'  
files = glob.glob(file_pattern)
```

```python
for filename in files:
    x = []
    y = []

    with open(filename, 'r') as file:
        lines = file.readlines() 
        for i in range(0, len(lines), 2):
            x.append(float(lines[i].strip())) 
            if i + 1 < len(lines):
                y.append(float(lines[i + 1].strip())) 

    
   
```

```python
    plt.scatter(x, y, marker='o', s=10)  
    plt.title(f'Бинарный поиск')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show() 
    plt.savefig('бинарный поиск.png')
```
